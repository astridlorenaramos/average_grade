We import pandas library

```python
import pandas as pd
```

We create a dictionary with every column of the data set, every key is a column name

```python
data = {
    "Student": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Math": [85, 78, 92, 75, 88],
    "Science": [90, 82, 88, 80, 92]
}
```

We need to change the dictionary to a pandas DataFrame 

```python
df = pd.DataFrame(data)
```

We calculate the average of every column. We access to every column of the DF by puting its name (math and science) and we add the operation to calculate the average by adding .mean()

```python
average_math = df["Math"].mean()
average_science = df["Science"].mean()
```

We do the same to find the highest grade in each subject, but this time using idxmax()

```python
highest_math = df.loc[df["Math"].idxmax()]
highest_science = df.loc[df["Science"].idxmax()]
```

We print the results by calling every var 

```python
print("Average Math:", average_math)
print("Average Science:", average_science)
print("Highest math grade:", highest_math["Student"], highest_math["Math"])
print("Highest Science grade:", highest_science["Student"], highest_science["Science"])
```
